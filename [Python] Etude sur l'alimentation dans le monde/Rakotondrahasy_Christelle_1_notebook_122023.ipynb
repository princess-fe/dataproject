```python
# Mettre en forme les données dans "Années" 
sous_nutrition['Année'] =  sous_nutrition['Année'].str.slice(5,9)
sous_nutrition['Année']=sous_nutrition['Année'].astype('int')
sous_nutrition['Année'] = sous_nutrition['Année'] - 1
```

```python
# Il faut tout d'abord faire une jointure entre la table population et la table sous nutrition, en ciblant l'année 2017
population_sous_nutrition = pd.merge (population , sous_nutrition , on = ["Zone", "Année"])
population_sous_nutrition = population_sous_nutrition[population_sous_nutrition.Année == 2017 ] 
```

```python
#Affichage du dataset
display(population_sous_nutrition)
```

```python
#Calcul et affichage du nombre de personnes en état de sous nutrition
proportion_sous_nutrition = (population_sous_nutrition['sous_nutrition'].sum() / population_sous_nutrition['Population'].sum())*100
print('Proportion de personnes en sous nutrition : {:.2f} %'.format(proportion_sous_nutrition))
```

```python
#Nombre de population en sous-nutrition en 2017
print('Nombre de population mondiale : {:.0f}'.format(population_sous_nutrition['Population'].sum()))
print('Nombre de personnes en sous nutrition : {:.0f}'.format(population_sous_nutrition['sous_nutrition'].sum()))
 
```

```python
#Pie chart : Proportion de personnes en état de sous-nutrition en 2017
#Récupération des données
proportion_pers =[proportion_sous_nutrition,(100-proportion_sous_nutrition)]
labels = ['Sous-nutrition', 'Autres']

#On isole la proportion des personnes en sous nutritions
explode = (0, 0.3)  
col_map = plt.get_cmap('Paired')

#Création du graph
fig, ax = plt.subplots()
ax.pie(proportion_pers, labels=None, explode=explode,autopct='%1.2f%%', shadow=True, startangle=90, colors=col_map.colors)
ax.axis('equal')
plt.legend(labels)
plt.title('Proportion de personnes en état de sous-nutrition en 2017\n', fontsize=14)

#exporter le graph
plt.savefig('C:\\Users\\CRFP4534\\OneDrive - orange.com\\Bureau\\Formation CFA - Data Analyst\\P4-Réalisez une étude de santé publique avec R ou Python\\img\\Top10 Pays_forte sous nutrition.png',bbox_inches='tight') 
plt.show()
```

<h3 style="margin: auto; padding: 20px; color: RGB(0,150,250); ">3.2 - Nombre théorique de personne qui pourrait être nourries</h3>

```python
#Combien mange en moyenne un être humain ? Source => Selon la FAO, c'est 2800 Calories ou bien entre 2900 et 2950 Calories
```

```python
#On commence par faire une jointure entre le data frame population et Dispo_alimentaire afin d'ajouter dans ce dernier la population
population_2017 = population[population.Année == 2017 ] 
nb_pers_nourrie = pd.merge (population_2017 , dispo_alimentaire , on = ["Zone"])
```

```python
#Affichage du nouveau dataframe
display(nb_pers_nourrie)
```

```python
#Création de la colonne dispo_kcal avec calcul des kcal disponibles mondialement
nb_pers_nourrie['dispo_kcal'] = nb_pers_nourrie['Disponibilité alimentaire (Kcal/personne/jour)'] * nb_pers_nourrie['Population']*365
```

```python
#Calcul du nombre d'humains pouvant être nourris
nb_theorique_pers_nourrie = nb_pers_nourrie['dispo_kcal'].sum()/ (2800 * 365)
print("Nombre d'humains pouvant être nourris : {:.0f} ".format(nb_theorique_pers_nourrie))
```

```python
#Bar chart : Nombre théorique de personnes qui pourraient être nourries en 2017
#Récupérer les données à afficher dans barh
label =  ["Nb théorique de personnes\n qui pourraient être nourries","Nb de population totale"]
population_2017 = population[population.Année == 2017 ] 
population_totale= population_2017['Population'].sum()
nb_population = [nb_theorique_pers_nourrie,population_totale]

#Ajuster la taille de l'image et obtenir la réf de couleur à utiliser
plt.figure(figsize=[2,1])
col_map = plt.get_cmap('Paired')

#Création de la graphique avec titre,légende des axes et étiquettes des données
fig, ax = plt.subplots()
hbars = plt.bar(label,nb_population , width= 0.5, color=col_map.colors)
#ax.set_xlabel("Proportion de personnes en état de sous-nutrition (%)", fontsize=15)
ax.set_title("Nombre théorique de personnes qui pourraient être nourries en 2017 \n", fontsize=20)
ax.bar_label(hbars, fmt='%.0f', label_type='center')
#ax.set_xlim(right=8000000000)

#exporter le graph
plt.savefig('C:\\Users\\CRFP4534\\OneDrive - orange.com\\Bureau\\Formation CFA - Data Analyst\\P4-Réalisez une étude de santé publique avec R ou Python\\img\\Nb théorique de personnes qui pourraient être nourries en 2017.png',bbox_inches='tight') 

#Afficher le graph
plt.show()
```

```python
#Calcul de la disponibilité totale
disponibilite = dispo_alimentaire.groupby('Zone')[['Aliments pour animaux',
                                                   'Disponibilité intérieure', 
                                                   'Exportations - Quantité', 
                                                   'Importations - Quantité', 
                                                   'Nourriture', 
                                                   'Pertes', 
                                                   'Production', 
                                                   'Semences', 
                                                   'Traitement', 
                                                   'Variation de stock',
                                                    'Autres Utilisations']].sum()
```

```python
#création d'une boucle for pour afficher les différentes valeurs en fonction des colonnes aliments pour animaux, pertes, nourritures, 
dispo_int = dispo_alimentaire['Disponibilité intérieure'].sum()

type_utilisation = ['Aliments pour animaux', 'Pertes', 'Nourriture','Semences', 'Traitement', 'Autres Utilisations']
for i in type_utilisation :
    disponibilite_mondiale = round((disponibilite[i].sum()/dispo_int)*100,2)
    print(f'disponibilité {i} : {disponibilite_mondiale}%')
```

```python
#Pie chart : Répartition de la disponibilité alimentaire
#Récupération des données
proportion_dispo =['13.24','4.61','49.51','1.57','22.38','8.78']
labels = ['Aliments pour animaux', 'Pertes', 'Nourriture','Semences', 'Traitement', 'Autres Utilisations']

#On isole la proportion des personnes en sous nutritions
explode = (0, 0.3,0,0,0,0)  

col_map = plt.get_cmap('Paired')
#Création du graph
fig, ax = plt.subplots()
ax.pie(proportion_dispo, labels=labels, explode=explode,autopct='%1.2f%%', shadow=True, startangle=90, colors=col_map.colors)
ax.axis('equal')
#plt.legend(labels)
plt.title('Répartition de la disponibilité alimentaire \n', fontsize=14)

#exporter le graph
plt.savefig('C:\\Users\\CRFP4534\\OneDrive - orange.com\\Bureau\\Formation CFA - Data Analyst\\P4-Réalisez une étude de santé publique avec R ou Python\\img\\Répartition de la disponibilité alimentaire.png',bbox_inches='tight') 
plt.show()
```
