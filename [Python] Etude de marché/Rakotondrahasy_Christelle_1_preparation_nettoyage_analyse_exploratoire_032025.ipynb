```python
# Lister toutes les catégories possibles pour les colonnes de type texte
colonnes_texte = disponibilite.select_dtypes(include=['object']).columns

for colonne in colonnes_texte:
    print(f"Catégories uniques pour la colonne '{colonne}':")
    print(disponibilite[colonne].unique())
    print("\n")
```

```python
# Supprimer les colonnes inutiles
colonnes_a_supprimer = ['Note', 'Code Élément', 'Code année']
disponibilite = disponibilite.drop(columns=colonnes_a_supprimer)
```

```python
# Filtrer les données pour ne garder que l'élément "disponibilite alimentaire"
disponibilite_alimentaire = disponibilite[disponibilite['Élément'] == 'Disponibilité alimentaire (Kcal)']
# Disponibilité alimentaire moyenne pour les produits animaux par année et par zone
disponibilite_moy = disponibilite_alimentaire.groupby(['Zone', 'Année'])['Valeur'].mean().reset_index()
```

```python
# Calcul des quartiles de la disponibilité alimentaire
Q1 = disponibilite_moy['Valeur'].quantile(0.25)  # Premier quartile (25%)
Q3 = disponibilite_moy['Valeur'].quantile(0.75)  # Troisième quartile (75%)
IQR = Q3 - Q1  # Intervalle interquartile

# Détection des valeurs aberrantes
borne_inferieure = Q1 - 1.5 * IQR
borne_superieure = Q3 + 1.5 * IQR

print(f"Borne inférieure : {borne_inferieure}")
print(f"Borne supérieure : {borne_superieure}")
```

```python
# Filtrer pour avoir les valeurs aberrantes
valeurs_aberrantes = disponibilite_moy[(disponibilite_moy['Valeur'] < borne_inferieure) | (disponibilite_moy['Valeur'] > borne_superieure)]

# Réorganiser les données pour avoir une colonne par année
valeurs_aberrantes= valeurs_aberrantes.pivot(index='Zone', columns='Année', values='Valeur')

# Calculer la moyenne des colonnes (années) pour chaque zone
valeurs_aberrantes['Moyenne'] = valeurs_aberrantes.mean(axis=1)

# Trier les zones par la moyenne (ordre décroissant)
valeurs_aberrantes = valeurs_aberrantes.sort_values(by='Moyenne', ascending=False)

# Afficher la table triée des valeurs aberrantes avec les moyennes
valeurs_aberrantes = valeurs_aberrantes[['Moyenne']]
print("Liste des Pays avec des valeurs aberrantes entre 2020 et 2022 (Disponibilité alimentaire)")
valeurs_aberrantes
```

```python
# Afficher le nombre de pays avec des valeurs aberrantes
print(f"Nombre de pays avec des valeurs aberrantes : {valeurs_aberrantes.shape[0]}")
```

```python
# Filtrer les données sans valeurs aberrantes
disponibilite_horsVA = disponibilite_moy[(disponibilite_moy['Valeur'] >= borne_inferieure) & (disponibilite_moy['Valeur'] <= borne_superieure)]

# Convertir la colonne 'Année' en type catégorique ou chaîne de caractères
disponibilite_horsVA.loc[:, 'Année'] = disponibilite_horsVA['Année'].astype(str)

# Créer un boxplot interactif horizontal avec les noms des zones en info-bulle
fig = px.box(
    disponibilite_horsVA,
    y='Année',  # Année devient l'axe vertical
    x='Valeur',  # Valeur devient l'axe horizontal
    color='Année',  # Optionnel : colorer les boxplots par année
    hover_data=['Zone'],  # Ajouter la colonne 'Zone' dans les info-bulles
    title='Distribution de la disponibilité alimentaire par année (hors valeurs aberrantes)'
)

# Personnaliser les axes et la taille de la figure
fig.update_layout(
    xaxis_title='Disponibilité alimentaire (en Kcal)',
    yaxis_title='Année',
    yaxis_tickangle=0,  # Pas de rotation pour les labels des années
    width=1000,  # Largeur de la figure
    height=600   # Hauteur de la figure
)

# Afficher le graphique interactif
fig.show()
```

```python
# Liste des Unités utilisées pour chaque Element
element_unite = disponibilite.groupby(['Élément', 'Unité']).size()
element_unite 
```

```python
# Convertir les valeurs 1000 t en tonnes et millions de kcal en kcal
disponibilite['Valeur'] = disponibilite.apply(
    lambda row: row['Valeur'] * 1000 if row['Unité'] == "1000 t" else (
        row['Valeur'] * 1000000 if row['Unité'] == "millions de kcal" else row['Valeur']
    ),
    axis=1
)
```

```python
# Pivoter les données selon la valeur de la colonne "Element" et année
disponibilite = disponibilite.pivot(index=['Code zone (ISO3)','Zone', 'Année','Produit'], columns='Élément', values='Valeur').reset_index()
```

```python
# Supprimer les colonnes inutiles pour les produits animaux ou qui présente une ambiguité
disponibilite = disponibilite.drop(columns=['Traitement', 'Autres utilisations (non alimentaire)', 'Résidus','Variation de stock','Aliments pour animaux'])
```

```python
# Filtrer sur les 'Viande de Volailles'
disponibilite_volaille = disponibilite[disponibilite['Produit'] == 'Viande de Volailles']

# Supprimer la colonne 'Produit'
disponibilite_volaille = disponibilite_volaille.drop(columns=['Produit'])
```

```python
# Dataframe disponibilité alimentaire des produits autres que la viande de volaille
disponibilite_autres = disponibilite[disponibilite['Produit'] != 'Viande de Volailles']

# Calculer la disponibilité alimentaire et l'importation des produits autres que la viande de volaille
disponibilite_autres = disponibilite_autres.groupby(['Zone', 'Année'])[['Disponibilité alimentaire (Kcal)', 'Importations - quantité']].sum().reset_index()

# Renommer les colonnes pour éviter les doublons
disponibilite_autres = disponibilite_autres.rename(columns={'Disponibilité alimentaire (Kcal)': 'Disponibilité alimentaire - Autres (Kcal)', 'Importations - quantité': 'Importations Autres - quantité'})
```

```python
# Concaténer les deux DataFrames sur les colonnes 'Zone' et 'Année'
disponibilite = pd.merge(
    disponibilite_volaille,
    disponibilite_autres,
    on=['Zone', 'Année'],
    how='outer'
)
```

```python
# Remplir les valeurs manquantes par 0
disponibilite = disponibilite.fillna(0)

# Afficher les premières lignes
disponibilite.head()
```

```python
# df = Disponibilite + Population
df = pd.merge(disponibilite, population, on=['Code zone (ISO3)', 'Année'], how='outer', suffixes=('_disponibilite', '_population'))

# df = df + PIB
df = pd.merge(df, pib, on=['Code zone (ISO3)', 'Année'], how='outer', suffixes=('', '_pib'))

# df = df + Stabilité Politique
df = pd.merge(df, politique, on=['Code zone (ISO3)', 'Année'], how='outer', suffixes=('', '_politique'))
```

```python
# Nombre de valeurs présentes dans chacune des colonnes
df.info()
```

```python
# Supprimer les colonnes inutiles
colonnes_a_supprimer = ['Code zone (ISO3)', 'Zone_disponibilite', 'Zone_population', 'Zone_politique']
df = df.drop(columns=colonnes_a_supprimer)

# Déplacer la colonne 'Zone' à la première position
df = df[['Zone'] + [col for col in df.columns if col != 'Zone']]
```

```python
# Supprimer la colonne 'Année'
df = df.drop(columns=['Année'])

# Moyenne annuelle des valeurs par pays et par produit
df = df.groupby(['Zone']).mean().reset_index()
```

```python
# Supprimer les lignes où la colonne 'Zone' est vide
df = df[df['Zone'].notna()]
```
